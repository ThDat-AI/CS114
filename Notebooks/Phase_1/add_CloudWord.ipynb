```python
import pandas as pd
import matplotlib.pyplot as plt

from wordcloud import WordCloud
```

```python
train_path = "/home/thanhf/Documents/Projects/cs114/Data/train.csv"
test_path = "/home/thanhf/Documents/Projects/cs114/Data/test.csv"
```

```python
df = pd.read_csv(train_path)
df.head()
```

```python
df.columns
```

```python
def plot_wordcloud(text, title):
    wc = WordCloud(
        width=800,
        height=400,
        background_color='white'
    ).generate(text)

    plt.figure(figsize=(10, 5))
    plt.imshow(wc, interpolation='bilinear')
    plt.title(title)
    plt.axis('off')
    plt.show()
```

```python
label_map = {
    1: "World",
    2: "Sports",
    3: "Business",
    4: "Sci/Tech"
}

# Gộp Title + Description trực tiếp
df['text'] = df['Title'] + " " + df['Description']

for label, name in label_map.items():
    text = " ".join(df[df['Class Index'] == label]['text'])
    plot_wordcloud(text, f"Word Cloud - {name}")
```

```python
label_map = {
    1: "World",
    2: "Sports",
    3: "Business",
    4: "Sci/Tech"
}

plt.figure(figsize=(14, 10))

for i, (label, name) in enumerate(label_map.items(), 1):
    text = " ".join(df[df['Class Index'] == label]['text'])
    
    wc = WordCloud(
        width=800,
        height=400,
        background_color='white'
    ).generate(text)
    
    plt.subplot(2, 2, i)
    plt.imshow(wc, interpolation='bilinear')
    plt.title(name)
    plt.axis('off')

plt.tight_layout()
plt.show()
```
